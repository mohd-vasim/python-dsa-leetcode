```python
# 8. Maximum current_diff between consecutive elements in a list

lst = [1, 7, 3, 10, 5]

def check_max_consecutive(lst: list):
    """Check max consecutive"""

    # Edge case
    if len(lst) < 2:
        return 0

    max_diff = 0

    for i in range(1, len(lst)):

        current_diff = abs(lst[i] - lst[i - 1])

        if current_diff > max_diff:

            max_diff = current_diff

    return max_diff

print(check_max_consecutive(lst))

print(check_max_consecutive([10, 11, 15, 3]))
```

This logic pattern is called **"Two-Pointer" or "Sliding Window"** approach, more specifically:

- **Two-Pointer Technique**: You're comparing two adjacent elements (`lst[i]` and `lst[i-1]`) as you iterate through the list
- **Linear/Sequential Traversal**: Moving through the list one step at a time, comparing consecutive pairs

In this case, it's examining **adjacent elements** to find the maximum difference between them. This is a common pattern used for:

- Comparing consecutive elements
- Finding gaps or differences between nearby items
- Processing paired data points

**Other names for similar patterns:**
- **Pairwise comparison**: Comparing each pair of consecutive elements
- **Window of size 2**: A minimal sliding window that only looks at two elements at a time

This approach is efficient with **O(n) time complexity** since you iterate through the list once, making only one pass with constant operations per iteration.

```python
# 9. Merge two sorted list

list1 = [1, 3, 5]
list2 = [2, 4, 6]

# Options:
# 1. merge two lists and use sorted function
# 2. 

def merge_two_sorted_lists(list1: list, list2: list):
    """Merge two sorted lists"""

    # Two pointers
    i, j = 0, 0
    new_list = []

    # Traverse until i or j as same as lists length
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            new_list.append(list1[i])

            # Update i for next comparison
            i += 1

        else:
            new_list.append(list2[j])

            # Update j for next comparison
            j += 1
        
    # If lists have different length
    while i < len(list1):
        new_list.append(list1[i])
        i += 1

    while j < len(list2):
        new_list.append(list2[j])
        j += 1
    
    return new_list

print(merge_two_sorted_lists(list1, list2))
```

```python
def rotate_list(lst, k):
    """Rotate a list"""

    if not lst:
        return []

    new_list = []
    n = len(lst)
    k %= n   # fix 1

    i = n - k

    while i < n:
        new_list.append(lst[i])
        i += 1

    i = 0
    while i < n - k:
        new_list.append(lst[i])
        i += 1

    return new_list

print(rotate_list(lst=[1, 2, 3, 4, 5], k=2))
print(rotate_list(lst=[10, 20, 30, 40, 50], k=3))
```

```python
# Another approach

def rotate_list(lst, k):
    n = len(lst)
    if n == 0:
        return []
    
    k %= n
    res = [0]*n
    
    for i in range(n):
        res[(i+k) % n] = lst[i]
    
    return res

print(rotate_list(lst=[1, 2, 3, 4, 5], k=2))
print(rotate_list(lst=[10, 20, 30, 40, 50], k=3))
```


### 11. Merge 2 List into Dictionary
Merge Lists to Dictionary

Design a Python function named merge_lists_to_dictionary to merge two lists into a dictionary where elements from the first list act as keys and elements from the second list act as values.

**Parameters:**

keys (List): A list of keys.

values (List): A list of values.

**Returns:**

A dictionary containing merged key-value pairs.

**Example:**

Input: keys = ['a', 'b', 'c'], values = [1, 2, 3]
Output: {'a': 1, 'b': 2, 'c': 3}

Input: keys = ['x', 'y', 'z'], values = [10, 20, 30]
Output: {'x': 10, 'y': 20, 'z': 30}

```python
keys = ['a', 'b', 'c']
values = ['x', 'y', 'z']

def merge_lists_to_dictionary(lst1, lst2):
    """Merge lists to a dictionary"""

    if not len(lst1) == len(lst2):
        return "Length of boths lists should be same"
    
    dictionary_from_lists = {}
    for i in range(len(lst1)):
        dictionary_from_lists[lst1[i]] = lst2[i]
    return dictionary_from_lists

print(merge_lists_to_dictionary(keys, values))
```

### 12. Merge Multiple Dictionaries
Merge Three Dictionaries

Design a Python function named merge_three_dictionaries to merge exactly three dictionaries into one.

**Parameters:**

dict1 (Dictionary): The first dictionary to be merged.

dict2 (Dictionary): The second dictionary to be merged.

dict3 (Dictionary): The third dictionary to be merged.

**Returns:**

A single dictionary containing all key-value pairs from the three input dictionaries.

**Example:**

Input: ({'a': 1, 'b': 2}, {'c': 3, 'd': 4}, {'e': 5, 'f': 6})
Output: {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5, 'f': 6}

Input: ({'x': 10, 'y': 20}, {'z': 30}, {'a': 40, 'b': 50})
Output: {'x': 10, 'y': 20, 'z': 30, 'a': 40, 'b': 50}

```python
def merge_mult_dicts(dicts: tuple[dict]):
    """Merge multiple dicts"""

    if not dicts:
        return "No dicts are available to merge"

    result_dict = {}
    for element in dicts:
        result_dict.update(element)

    return result_dict

dicts = ({'a': 1, 'b': 2}, {'c': 3, 'd': 4}, {'e': 5, 'f': 6})

print(merge_mult_dicts(dicts))
```

### 13. Words Frequency in a Sentence
Count Word Frequency

Design a Python function named count_word_frequency to count the frequency of words in a sentence and store the counts in a dictionary.

**Parameters:**

sentence (str): The input sentence where you need to count the frequency of each word.

**Returns:**

A dictionary where the keys are words from the sentence and the values are their corresponding frequencies.

**Example:**

Input: "hello world hello"
Output: {'hello': 2, 'world': 1}

Input: "the quick brown fox jumps over the lazy dog"
Output: {'the': 2, 'quick': 1, 'brown': 1, 'fox': 1, 'jumps': 1, 'over': 1, 'lazy': 1, 'dog': 1} 


```python
def count_word_frequency(string: str):
    """Count word frequencdy"""
    splitted_words = string.split()
    words_count = {}
    for word in splitted_words:
        if not word in words_count:
            words_count[word] = 1
        else:
            words_count[word] += 1

    return words_count

sample_input = "hello world hello"
print(count_word_frequency(sample_input))
```

```python
def check_palindrome(test_case):
    """Check palindrome"""

    n = len(test_case)
    i, j, mid = 0, n - 1, n // 2

    while i < mid:
        if not test_case[i] == test_case[j]:
            return "Test case is not palindrome"
        
        i += 1
        j -= 1
    
    return "Test case is palindrome"

test1 = (1, 2, 3, 2, 1)
print(check_palindrome(test1))
```

### 15. Merge Dictionaries with Common Keys
Problem Description

Merge Dictionaries with Overlapping Keys

Design a Python function named merge_dicts_with_overlapping_keys that merges multiple dictionaries into a single dictionary. If a key appears in more than one dictionary, sum up their values.

**Parameters:**

dicts (list): A list of dictionaries where keys might overlap.

**Returns:**

A single dictionary where values for overlapping keys are summed.

**Example:**

Input: [{'a': 1, 'b': 2}, {'b': 3, 'c': 4}, {'c': 5, 'd': 6}]
Output: {'a': 1, 'b': 5, 'c': 9, 'd': 6}

Input: [{'x': 10, 'y': 20}, {'y': 30, 'z': 40}, {'z': 50, 'x': 60}]
Output: {'x': 70, 'y': 50, 'z': 90}

```python
def merge_dicts_with_overlapping_keys(test_case: list[dict[str, str]]):
    """Merge dicts with overlapping keys"""

    new_dict = {}
    for d in test_case:
        for key, value in d.items():
            if not key in new_dict:
                new_dict[key] = value

            else:
                new_dict[key]  += value
    
    return new_dict

test1 = [{'a': 1, 'b': 2}, {'b': 3, 'c': 4}, {'c': 5, 'd': 6}]
print(merge_dicts_with_overlapping_keys(test1))
```

```python
test2 = [{'x': 10, 'y': 20}, {'y': 30, 'z': 40}, {'z': 50, 'x': 60}]
print(merge_dicts_with_overlapping_keys(test2))
```
