### Factorial of a number using Recursion

````python
def factorial(n):
    """Calculate factorial"""

    if n == 1 or n == 0:
        return 1

    return n * factorial(n - 1)
````

````python
print(factorial(3))
print(3*2*1)
````

### Sum of N numbers using Recursion

````python
def sum_n(n):
    """Sum N natural numbers"""
    total = 0

    if n == 1:
        return 1

    total = n + sum_n(n-1)
    return total
````

````python
print(sum_n(3))
print(sum_n(4))
````

### Number of Digits using Recursion

````python
def num_digits(n):

    # base case
    if n < 10:
        return 1
    
    n_digits = 1 + num_digits(n / 10)

    return n_digits
````

````python
print(123/10)
print(12/10)
print(1/10)
````

````python
num_digits(123)
````

## 4. Fibonacci Series using Recursion

### Problem Description
You are given a non-negative integer `n`. Your task is to find the **nth Fibonacci number** using recursion.

### Fibonacci Sequence Definition
```
F(0) = 0
F(1) = 1
F(n) = F(n-1) + F(n-2)  for n >= 2
```

### Input
- A non-negative integer `n`, where `0 <= n <= 30`

### Output
- An integer representing the nth Fibonacci number

### Examples

| Input | Output | Explanation |
|-------|--------|-------------|
| `n = 5` | `5` | Sequence: `0, 1, 1, 2, 3, 5` |
| `n = 0` | `0` | `F(0) = 0` |

---

n = 5

Small problem:
0, 1 -> 0 + 1 = 1
sequence = 0, 1, 1

0, 1, 1 -> 1 + 1 = 2
sequence = 0, 1, 1, 2

0, 1, 1, 2 -> 1 + 2 = 3
sequence = 0, 1, 1, 2, 3

0, 1, 1, 2, 3 -> 2 + 3 = 5
sequence = 0, 1, 1, 2, 3, 5



````python
def fibonacci(n):
    """Find fibonacci series"""
    if n == 0:
        return 0

    if n == 1:
        return 1

    last = fibonacci(n-1)
    second_last = fibonacci(n-2)

    addition = last + second_last

    return addition

````

````python
print(fibonacci(6))
````

## 5. Print 1 to N using Recursion

### Problem Description
You are given a positive integer `n`. Your task is to **return a list of integers from `1` to `n`** using recursion.

### Input
- A positive integer `n`, where `1 <= n <= 1000`

### Output
- A list of integers from `1` to `n`

### Examples

| Input | Output |
|-------|--------|
| `n = 5` | `[1, 2, 3, 4, 5]` |
| `n = 3` | `[1, 2, 3]` |

---

````python
def print_to_n(n):
    """Print from 1 to N"""

    if n == 0:
        return []

    current_list = [n]
    new_item = print_to_n(n - 1)

    # Append
    whole_list = new_item + current_list

    return whole_list
````

````python
print(print_to_n(5))
````
