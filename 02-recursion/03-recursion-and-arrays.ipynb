## Sum of an Array using Recursion

### Problem Description
You are given an array of integers. Your task is to calculate and return the sum of all elements in the array using recursion.

### Input
- An array of integers.

### Output
- An integer representing the sum of the array elements.

### Examples
- Input: `arr = [1, 2, 3, 4, 5]` → Output: `15`
- Input: `arr = [10, -2, 3, 5]` → Output: `16`

base case:
- if size(n) == 0:
    return 0

if size(n) == 1:
    return n

recursive case:
- sum_array(list[1:])
- list[i] + sum_array(list[1:])

```python
def sum_array(array):
    """Sum of elements in array"""
    if len(array) == 0:
        return 0

    if len(array) == 0:
        return array[0]

    ith_element = sum_array(array[1:])

    ans = ith_element + array[0]

    return ans
```

```python
print(sum_array([1, 2, 3, 4]))
```

```python
print(sum_array([]))
```

## First Index of an Element using Recursion

### Problem Description
You are given an array and a target element. Your task is to find the first index where the array contains that element using recursion. If the element is not found, return `-1`.

### Input
- An array `arr` of integers.
- An integer `element`.

### Output
- An integer representing the first index where `element` is found in `arr`, or `-1` if it is not found.

### Examples
- Input: `arr = [1, 2, 3, 2, 4, 2]`, `element = 2` → Output: `1`
- Input: `arr = [5, 6, 7, 8]`, `element = 10` → Output: `-1`
